**Multiplier Filter**

```python
for i in range(1, 31):
  if(i % 3 == 0 and i % 5 == 0):
    print(i)
```

**Natural Accumulator**

```python
total = 0
for i in range(1, 51):
  total += i
print(total)
```

**Reverse Countdown**

```python
for i in range(10, 0, -1):
  print(i)
```

**Average Calculator**

```python
num = int(input("Enter a number: "))
count = 0
sum = 0
for i in range(1, num+1):
  if i % 2 != 0:
    sum += i
    count += 1
print(f"The sum is {sum}")
print(f"The count is {count}")
average = sum / count
print(f"The average is {average}")
```

**Boundary Checker**

```python
L = int(input("Enter a number: "))
for i in range(1, L+1):
  sq = i*i
  if sq > L:
    break
  print(sq)
```

**Foctorial**

```python
num = int(input("Enter number :"))
fact = 1
for i in range(1, num+1):
  fact *= i
print(fact)
```
