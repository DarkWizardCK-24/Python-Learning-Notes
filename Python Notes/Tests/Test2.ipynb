**Balanced Alternator**

```python
for i in range(1, 21):
  if i % 3 == 0 and i % 5 == 0:
    print("Combo")
  elif i % 3 == 0:
    print("Triple")
  elif i % 5 == 0:
    print("Quint")
  else:
    print(i)
```

**Step Sipping Multipliers**

```python
num = int(input("Enter a number: "))
for i in range(2, 11, 2):
  print(f"{num} X {i} = {num*i}")
```

```python
num = int(input("Enter a number: "))
for i in range(1, 11):
  if i % 2 == 0:
    print(f"{num} X {i} = {num*i}")
```

**Dual Track Accumullator**

```python
sum = 0
prod = 1
for i in range (1, 11):
  if i % 2 != 0:
    prod *= i
  else:
    sum += i
print(prod)
print(sum)
```

**Negative Multiples**

```python
for i in range (-1, -31, -1):
  if i % 4 == 0:
    print(i)
```

**Factor Factorial**

```python
num = int(input("Enter a number: "))
prod = 1
for i in range (1, num):
  if num % i == 0:
    prod *= i
    print(i)
print(prod)
```

**Target Value Sentinal**

```python
num = int(input("Enter a number: "))
match = 0
for i in range (1, 51):
  if i % num == 0:
    match += 1
    print(i)
print(f"Total match : {match}")
```
