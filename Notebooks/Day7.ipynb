**Reverse Countdown**

```python
for i in range(10, 0, -1):
    print(i)
```

**Natural Accumulator**

```python
total = 0
for i in range (1, 51):
   total += i
print(total)
```

**Multiples Filter**

```python
for i in range(1, 31):
    if i % 3 == 0 and i % 5 == 0:
        print(i)
```

```python
L = int(input("Enter a limit number L: "))

for i in range(1, L + 1):
    square = i * i

    if square > L:
        break
    print(square)
```
